```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np
```

```python
headers={
    'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'
} 

webpage = {}

for i in range(1,6):
    url = f'https://www.ambitionbox.com/list-of-companies?page={i}'
    webpage[f'page{i}'] = requests.get(url, headers=headers).text
```

## TO FIND OUT NAMES OF THE COMPANIES

```python
for page_no, html in webpage.items():
    soup = BeautifulSoup(html, 'html.parser')
    companies = soup.find_all('h2', class_='companyCardWrapper__companyName')
    print(f"\n--- Page {page_no} ---", len(companies))

    for company in companies:
        print(company.text.strip())
```

## TO FIND OUT THE RATINGS

```python
for page_no, html in webpage.items():
    soup = BeautifulSoup(html, 'html.parser')
    rating_div = soup.find_all("div", class_="rating_text rating_text--md")
    print(f"\n--- Page {page_no} ---", len(rating_div))

    for rat in rating_div:
        rating = rat.find("div").text.strip()
        print(rating)
```

## TO FIND OUT THE NUMBER OF REVIEWS

```python
for page_no, html in webpage.items():
    soup = BeautifulSoup(html, 'html.parser')
    rating_counts = soup.find_all("span", class_="companyCardWrapper__companyRatingCount")
    print(f"\n--- Page {page_no} ---", len(rating_counts))

    for count in rating_counts:
        text = count.text.strip()
        rating_count = text[1:-1]
        print(rating_count)
```

## CONSIDERING THE WHOLE CONTAINER

```python
from concurrent.futures import ThreadPoolExecutor, as_completed
import time
import random
```

```python
session = requests.Session()

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.ambitionbox.com/"
}

# Lists
name = []
ratings = []
reviews = []
sectors = []
locations = []
highly_rated_for = []
critically_rated_for = []

page_num = 1

while True:

    url = f"https://www.ambitionbox.com/list-of-companies?page={page_num}"

    try:
        response = session.get(url, headers=headers, timeout=10)

        soup = BeautifulSoup(response.text, "html.parser")

        companies = soup.find_all("div", class_="companyCardWrapper")

        # Stop automatically if no companies found
        if len(companies) == 0:
            print("No more pages found. Stopping.")
            break

        print(f"Scraping page {page_num}...")

        for container in companies:

            # Name
            name_tag = container.find("h2", class_="companyCardWrapper__companyName")
            name.append(name_tag.text.strip() if name_tag else np.nan)

            # Rating
            rating_div = container.find("div", class_="rating_text rating_text--md")
            rating_inner = rating_div.find("div") if rating_div else None
            ratings.append(rating_inner.text.strip() if rating_inner else np.nan)

            # Reviews
            review_tag = container.find("span", class_="companyCardWrapper__ActionCount")
            reviews.append(review_tag.text.strip() if review_tag else np.nan)

            # Sector and location
            profile_tag = container.find("span", class_="companyCardWrapper__interLinking")

            industry = np.nan
            location = np.nan

            if profile_tag:
                parts = profile_tag.text.strip().split("|")

                if len(parts) == 2:
                    industry = parts[0].strip()
                    location = parts[1].strip()

                elif len(parts) == 1:
                    text = parts[0].lower()

                    cities = ["bangalore", "mumbai", "delhi", "gurgaon",
                              "noida", "hyderabad", "chennai", "pune"]

                    if any(city in text for city in cities):
                        location = parts[0].strip()
                    else:
                        industry = parts[0].strip()

            sectors.append(industry)
            locations.append(location)

            # Highly and critically rated
            high_value = np.nan
            critical_value = np.nan

            rating_blocks = container.find_all(
                "div", class_="companyCardWrapper__ratingHeader"
            )

            for block in rating_blocks:

                heading_tag = block.find("span")

                if heading_tag:

                    heading = heading_tag.text.lower()

                    value_tag = block.find_next_sibling(
                        "span",
                        class_="companyCardWrapper__ratingValues"
                    )

                    value = value_tag.text.strip() if value_tag else np.nan

                    if "highly" in heading:
                        high_value = value

                    elif "critical" in heading:
                        critical_value = value

            highly_rated_for.append(high_value)
            critically_rated_for.append(critical_value)

        page_num += 1

        # Small random delay (fast but safe)
        time.sleep(random.uniform(0.2, 0.4))

    except Exception as e:
        print("Error:", e)
        break
```

```python
final_df = {
    'company_name': name,
    'ratings':ratings,
    'reviews':reviews,
    'Industry':sectors,
    'locations':locations,
    'highly_rated_aspects':highly_rated_for,
    'critically_rated_aspects':critically_rated_for
}
```

```python
company_dataframe = pd.DataFrame(final_df)
```
